**TASK 2.1 (10 Points):** Lets begin by fixing the Birthday column. You may notice that the dates are in inconsistent format!

- Pandas internally represent dates in 'datetime64' format which is a form of ISO 8601. Basically, dates are represented as yyyy-mm-dd.


- Fix all the dates in the Birthday Column of `profiles_df` dataframe (Do not use other variable name).
- Display the `profiles_df` dataframe (only top 5 rows) after you are done.
- Note that the only **2 formats** the dates are in is either **yyyy/dd/mm** or **mm/dd/yyyy**;
- You want to represent them as **yyyy-mm-dd**

- **EXAMPLE**: In the “Profiles” dataset, dates like *2000/10/12* and *10/12/2000* follow different formats—one as *December 10th, 2000*, and the other as *October 12th, 2000*. We aim to standardize them to *2000-12-10* and *2000-10-12*, respectively.

**HINT:** some datetime functions might make things easier.
- https://www.geeksforgeeks.org/python-pandas-to_datetime/
- https://www.programiz.com/python-programming/datetime/strftime

**Outputs to be Graded**
- Variable(s): `profiles_df`
- Cell outputs


```python
# Write your code 
#Chanign th date format to y/m/d

def format_func(date):
    s = ""
    count = 0;
    while  True:
        if date[count] == '/' and count > 2:
            return pd.to_datetime(date, format = '%Y/%d/%m')
        elif date[count] == '/' and (count == 2 or count == 1):
            return pd.to_datetime(date, format = '%m/%d/%Y')
        else:
            count += 1
            

series = profiles_df['Birthday'].apply(format_func)
profiles_df['Birthday'] = series

profiles_df.head()
```

```python
# Write your code here

alaran_df = interrogation_df[interrogation_df['Name'] == 'Alan Turing']
encoded_message = alaran_df['Statement'].iloc[0]
encoded_message = encoded_message.replace('"', "")
#Split the whole setences into words
split_into_word = encoded_message.split("  ")
#split the space betwen the words
words = []
#make it into a single word
for word_with_single_space in split_into_word:
    words.append(word_with_single_space.split(" "))

#loop thrught each numebr and covert it into a letter
decoded_message = "" 
for word in words:
    for num in word:
        decoded_message += string.ascii_lowercase[int(num) - 1]
        
    
    decoded_message += " "
    
#strip any leading white spaces 
decoded_message = decoded_message.strip()
    
print({"Encoded Message:", encoded_message})
print({"Decoded Message:", decoded_message})
```

```python
# Write your code here
# Monk D Luffy67 statemnet Sum the individual #s in ticket numbers, look at the rightmost
# value,  0-2:B, 3-5:R , 6-7:G, 8-9:Y

def add_func(numbers):
    total = 0
    for num in str(numbers):
        total += int(num)
    last_digit = total % 10 #module 10 gives always the last digit
    if last_digit <= 2:
        return "Blue"
    elif last_digit > 2 and last_digit  <= 5:
        return "Red"
    elif last_digit == 6 or last_digit  == 7: #67 haha
        return "Green"
    else:
        return "Yellow"

ticket_color_series = show_df['Ticket_Number'].apply(add_func)

show_df['Ticket_Color'] = ticket_color_series


show_df.head()
```
